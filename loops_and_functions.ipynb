```python
#LOOPS
for i in range(2,9):
  print(i)
```

```python
for i in range(12,2,-3):
  print(i)
```

```python
s="mmdu"
for i in s:
  print(i)
```

```python
#while
i=0
while i<5:
  print(i)
  i=i+1
```

```python
#break
for i in range(10):
  print(i)
  if i==5:
    break
```

```python
#continue
for i in range(5):
  if i==2:
    continue
  print(i)
```

```python
#nested loops
for i in range(3):
  for j in range(2):
    print(f"i:{i} and j:{j}")
```

```python
#calculate the sum of first n natural numbers
n=int(input("enter the first N natural numbers:"))
sum=0
for i in range(n):
  sum=sum+i
  i+=1
print(sum)
```

```python
#pattern printing
for i in range(4):
  for j in range(4):
    print("*",end=" " )
  print()
```

```python
for i in range(5):
  for j in range(i+1):
    print("*", end="")
  print()
```

```python
rows = 5
for i in range(rows, 0, -1):
    for j in range(0, i):
        print('*', end='')
    print()  # Moves to the next line
```

```python
#FUNCTIONS

#function definition
def greet():
  print("hello")
```

```python
#function call
for i in range(10):
  greet()
```

```python
#types of functions
#IN BUILD: len(),print(),input,sum,max,min
#USER DEFINED
```

```python
#local and global values in a function
b=4#global
def fun():
  a=10#local
  print(a)
print(fun())
```

```python
#parameters in a function
def add(a,b):
  print(a+b)
add(11,24)
```

```python
#return statement
def add(a,b):
  return a+b,a-b,a*b
add(11,24)
```

```python
#default parameter
def fun(name="isha"):
  print(f"hello {name}")
fun("omkar")
```

```python
#keywords and positional arguments
#POSITIONAL
def student(name,age):
  print(f" {name} is {age} years old")
student(55,"isha")
#KEYWORD
student(age=30,name="omkar")
```

```python
#WRITE A PROGRAM TO FIND MAXIMUM OF THREE NUMBERS
def maximum(a,b,c):
  if a>b and a>c:
    print("A is greater number")
  elif b>c and b>a:
    print("b is greater number")
  else:
    print("c is the greater number")
maximum(99,88,105)



```

```python
#write a program of multiplication table of a number using a function
def table(a):
  product=0
  for i in range(1,11):
    product=i*a
    i+=1
    print(product)
table(11)
```
